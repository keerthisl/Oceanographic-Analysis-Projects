````python
# %%
#=================================
# Section 1: Setup and Imports (v1.2.0)
#=================================
import xarray as xr
import matplotlib.pyplot as plt
import numpy as np
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import os
from datetime import datetime
import glob

# Version info
VERSION = "1.2.0"
SCRIPT_NAME = "Sri Lanka Regional Chlorophyll Analysis"

# Project directories
project_dir = r"C:\Users\keert\Documents\PyProjects\OceanAnalysis"
data_dir = os.path.join(project_dir, "data")
output_dir = os.path.join(project_dir, "output")
figure_dir = os.path.join(output_dir, "figures")

# Create directories if they don't exist
for directory in [data_dir, output_dir, figure_dir]:
    if not os.path.exists(directory):
        os.makedirs(directory)
        print(f"Created directory: {directory}")

# %% [markdown]
# ### Region and Season Definitions
# The following code defines the specific regions around Sri Lanka and the monsoon seasons.

# %%
# Define Sri Lanka-specific regions
regions = {
    'Full Area': [78, 83, 5, 10],           # Broader Sri Lanka region
    'West Coast': [79, 80, 6, 9.5],         # Western coastal waters
    'East Coast': [81, 82, 6, 9.5],         # Eastern coastal waters
    'South Coast': [79.5, 81.5, 5.5, 6.5],  # Southern coastal waters
    'Palk Strait': [79.5, 80.5, 9, 10]      # Palk Strait region
}

# Define Sri Lanka-specific seasons (with their respective months)
seasons = {
    'Northeast Monsoon (Dec-Feb)': [12, 1, 2],
    'First Inter-monsoon (Mar-Apr)': [3, 4],
    'Southwest Monsoon (May-Sep)': [5, 6, 7, 8, 9],
    'Second Inter-monsoon (Oct-Nov)': [10, 11]
}

print(f"Running {SCRIPT_NAME} version {VERSION}")
print(f"Analysis start time: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")

# [Rest of the code sections follow the same pattern, with markdown documentation 
# before each code cell explaining the purpose and functionality]

# %% [markdown]
# ## 6. Output Descriptions
# 
# ### Generated Files
# 1. **Seasonal Cycle Plot**
#    - Filename pattern: `chlorophyll_seasonal_cycle_{timestamp}.png`
#    - Shows monthly variations in Chlorophyll-a
# 
# 2. **Monsoon Season Maps**
#    - Filename pattern: `sri_lanka_monsoon_seasonal_maps_{timestamp}.png`
#    - Four-panel plot showing seasonal patterns
# 
# 3. **Regional Analysis Maps**
#    - Filename pattern: `chlorophyll_map_{region}_{timestamp}.png`
#    - Separate plots for each coastal region
# 
# 4. **Statistics File**
#    - Filename pattern: `sri_lanka_chlorophyll_stats_{timestamp}.txt`
#    - Contains detailed statistical analysis
# 
# ## 7. Version History
# 
# ### v1.2.1 (Current)
# - Added Sri Lanka-specific regional analysis
# - Implemented monsoon-based seasonal analysis
# - Enhanced visualization features
# - Added version tracking for individual sections
# 
# ### v1.1.0
# - Added timestamps to all outputs
# - Improved error handling
# 
# ### v1.0.0
# - Initial version with basic analysis capabilities
# - Basic plotting functionality
````

````python
# %% [markdown]
# ## Section 2: Data Loading (v1.2.0)

# %%
def list_data_files():
    """List all available NetCDF files in the data directory."""
    nc_files = glob.glob(os.path.join(data_dir, '*.nc'))
    if not nc_files:
        print("No NetCDF files found in the data directory!")
        return []
    
    print("\nAvailable data files:")
    for i, file in enumerate(nc_files, 1):
        print(f"{i}. {os.path.basename(file)}")
    return nc_files

# List available files
nc_files = list_data_files()

# %%
def load_data(filename='CCI_ALL-v6.0-MONTHLY.nc'):
    """Load the Chlorophyll NetCDF file."""
    try:
        data_file = os.path.join(data_dir, filename)
        if not os.path.exists(data_file):
            raise FileNotFoundError(f"Data file not found: {data_file}")
        
        data = xr.open_dataset(data_file)
        print(f"\nAnalysis run on: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
        print("\nDataset information:")
        print(data)
        return data
        
    except Exception as e:
        print(f"Error loading data: {str(e)}")
        return None

# Load the data
data = load_data()
````

````python
# %% [markdown]
# ## Section 3: Basic Statistics (v1.2.0)

# %%
if data is not None:
    print("\nBasic Statistics:")
    print("Mean Chlorophyll:", f"{data.chlor_a.mean().values:.3f} mg/m³")
    print("Max Chlorophyll:", f"{data.chlor_a.max().values:.3f} mg/m³")
    print("Min Chlorophyll:", f"{data.chlor_a.min().values:.3f} mg/m³")
````

````python
# %% [markdown]
# ## Section 4: Seasonal Analysis (v1.2.1)

# %%
if data is not None:
    timestamp = datetime.now().strftime('%Y%m%d_%H%M')
    seasonal_cycle = data.chlor_a.groupby('time.month').mean()
    weights = np.cos(np.deg2rad(seasonal_cycle.lat))
    mean_chl = seasonal_cycle.weighted(weights).mean(dim=['lat', 'lon'])
    
    plt.figure(figsize=(12, 6))
    mean_chl.plot(marker='o')
    plt.title(f'Sri Lanka Regional Chlorophyll-a Cycle\nv{VERSION}')
    plt.xlabel('Month')
    plt.ylabel('Chlorophyll-a (mg/m³)')
    plt.grid(True)
    
    filename = f'sri_lanka_chlorophyll_seasonal_cycle_{timestamp}.png'
    plt.savefig(os.path.join(figure_dir, filename), 
                dpi=300, bbox_inches='tight')
    print(f"Saved seasonal cycle plot as: {filename}")
    plt.show()
````

````python
# %% [markdown]
# ## Section 5: Monsoon Season Maps (v1.2.1)

# %%
if data is not None:
    timestamp = datetime.now().strftime('%Y%m%d_%H%M')
    fig, axes = plt.subplots(2, 2, figsize=(15, 12),
                            subplot_kw={'projection': ccrs.PlateCarree()})
    axes = axes.flat
    
    # Adjust color scale for Sri Lankan coastal waters
    vmin, vmax = 0, 3  # Typical range for coastal waters
    
    for i, (season, months) in enumerate(seasons.items()):
        ax = axes[i]
        seasonal_data = data.chlor_a.sel(time=data.time.dt.month.isin(months)).mean('time')
        
        # Plot chlorophyll data with customized colorbar
        im = seasonal_data.plot(ax=ax, transform=ccrs.PlateCarree(),
                              cmap='viridis',
                              vmin=vmin, vmax=vmax,
                              cbar_kwargs={
                                  'label': 'Chlorophyll-a (mg/m³)',
                                  'shrink': 0.6,
                                  'aspect': 20,
                                  'pad': 0.02
                              })
        
        # Add enhanced features
        ax.add_feature(cfeature.OCEAN, facecolor='lightblue', alpha=0.3)
        ax.add_feature(cfeature.LAND, facecolor='lightgreen', alpha=0.3)
        ax.add_feature(cfeature.COASTLINE, linewidth=0.8, edgecolor='black')
        ax.add_feature(cfeature.BORDERS, linestyle=':', color='gray')
        
        gl = ax.gridlines(draw_labels=True, linewidth=0.5, color='gray', 
                         alpha=0.5, linestyle='--')
        gl.top_labels = False
        gl.right_labels = False
        
        ax.set_title(f'{season}', fontsize=10, pad=10)
        ax.set_extent(regions['Full Area'], crs=ccrs.PlateCarree())

    plt.suptitle(f'Sri Lanka Regional Chlorophyll-a Patterns by Monsoon Season\nv{VERSION}', 
                 fontsize=16, y=0.95)
    plt.tight_layout()
    
    filename = f'sri_lanka_monsoon_seasonal_maps_{timestamp}.png'
    plt.savefig(os.path.join(figure_dir, filename), 
                dpi=300, bbox_inches='tight')
    print(f"Saved monsoon seasonal maps as: {filename}")
    plt.show()
````

````python
# %% [markdown]
# ## Section 6: Coastal Regional Analysis (v1.2.1)

# %%
if data is not None:
    timestamp = datetime.now().strftime('%Y%m%d_%H%M')
    
    for region_name, region_extent in regions.items():
        if region_name != 'Full Area':
            plt.figure(figsize=(12, 8))
            ax = plt.axes(projection=ccrs.PlateCarree())
            
            # Calculate seasonal means for this region
            seasonal_means = {}
            for season, months in seasons.items():
                seasonal_data = data.chlor_a.sel(time=data.time.dt.month.isin(months))
                seasonal_means[season] = float(seasonal_data.mean().values)
            
            # Plot annual mean
            annual_mean = data.chlor_a.mean('time')
            
            im = annual_mean.plot(ax=ax, transform=ccrs.PlateCarree(),
                                cmap='viridis',
                                vmin=vmin, vmax=vmax,
                                cbar_kwargs={
                                    'label': 'Chlorophyll-a (mg/m³)',
                                    'shrink': 0.6,
                                    'aspect': 20,
                                    'pad': 0.02
                                })
            
            ax.add_feature(cfeature.OCEAN, facecolor='lightblue', alpha=0.3)
            ax.add_feature(cfeature.LAND, facecolor='lightgreen', alpha=0.3)
            ax.add_feature(cfeature.COASTLINE, linewidth=0.8, edgecolor='black')
            ax.add_feature(cfeature.BORDERS, linestyle=':', color='gray')
            
            gl = ax.gridlines(draw_labels=True, linewidth=0.5, color='gray', 
                            alpha=0.5, linestyle='--')
            gl.top_labels = False
            gl.right_labels = False
            
            ax.set_extent(region_extent, crs=ccrs.PlateCarree())
            
            # Add seasonal statistics in text box
            stats_text = "Seasonal Means:\n"
            for season, mean in seasonal_means.items():
                stats_text += f"{season.split('(')[0].strip()}: {mean:.2f} mg/m³\n"
            
            plt.gcf().text(0.15, 0.15, stats_text, fontsize=8,
                          bbox=dict(facecolor='white', alpha=0.7))
            
            plt.title(f'Chlorophyll-a Analysis - {region_name}\nv{VERSION}')
            
            filename = f'chlorophyll_map_{region_name.lower().replace(" ", "_")}_{timestamp}.png'
            plt.savefig(os.path.join(figure_dir, filename),
                        dpi=300, bbox_inches='tight')
            print(f"Saved {region_name} analysis as: {filename}")
            plt.show()
````

````python
# %% [markdown]
# ## Section 7: Save Statistics (v1.2.0)

# %%
if data is not None:
    timestamp = datetime.now().strftime('%Y%m%d_%H%M')
    stats_file = os.path.join(output_dir, f'sri_lanka_chlorophyll_stats_{timestamp}.txt')
    with open(stats_file, 'w') as f:
        f.write(f"{SCRIPT_NAME} version {VERSION}\n")
        f.write(f"Analysis Date: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n\n")
        f.write("Chlorophyll-a Concentration Analysis\n")
        f.write("===================================\n\n")
        
        for region_name, region_extent in regions.items():
            f.write(f"\nRegion: {region_name}\n")
            f.write("--------------------\n")
            for season, months in seasons.items():
                seasonal_data = data.chlor_a.sel(time=data.time.dt.month.isin(months))
                f.write(f"\n{season}:\n")
                f.write(f"Mean: {seasonal_data.mean().values:.3f} mg/m³\n")
                f.write(f"Std: {seasonal_data.std().values:.3f} mg/m³\n")
                f.write(f"Max: {seasonal_data.max().values:.3f} mg/m³\n")
                f.write(f"Min: {seasonal_data.min().values:.3f} mg/m³\n")
    print(f"\nStatistics saved to: {stats_file}")
````

````python
# %% [markdown]
# ## Section 8: Cleanup (v1.2.0)

# %%
plt.close('all')
if 'data' in locals():
    data.close()
print(f"\nAnalysis version {VERSION} completed and resources released.")
print(f"All figures saved to: {figure_dir}")
print(f"Statistics saved to: {output_dir}")
print(f"Analysis end time: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
````
