````python
!pip install cartopy  # Install cartopy if not already present
import matplotlib.pyplot as plt
import cartopy.crs as ccrs

# Define the map configuration data
map_config = {
    "projection": "PlateCarree",
    "figure_size": (5, 3),
    "features": ["coastlines", "gridlines"],
    "title": "World Map with Cartopy"
}

# Display the configuration data
print("Map configuration data:")
for key, value in map_config.items():
    print(f"{key}: {value}")

# Create the map
plt.figure(figsize=map_config["figure_size"])
ax = plt.axes(projection=ccrs.PlateCarree())
ax.coastlines()
ax.gridlines()
plt.title(map_config["title"])
plt.show()
````
